```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# Load training data
train_df = pd.read_csv("data/train.csv")
```

```python
# Simulate "new" incoming data by taking a sample with a shift
# In real life this would be new data from production
new_df = train_df.copy()
new_df["GrLivArea"] = new_df["GrLivArea"] * 1.5  # simulate drift
```

```python
# Pick features to monitor
features_to_monitor = ["GrLivArea", "OverallQual", "TotalBsmtSF"]
```

```python
for feature in features_to_monitor:
    stat, p_value = stats.ks_2samp(train_df[feature], new_df[feature])
    status = "DRIFT DETECTED" if p_value < 0.05 else "OK"
    print(f"{feature}: p-value={round(p_value, 4)} → {status}")
```

```python
features = ['LotFrontage', 'LotArea', 'OverallCond']
first_half = train_df[features].copy().iloc[:int(len(train_df)/2)]
second_half = train_df[features].copy().iloc[int(len(train_df)/2):]
```

```python
for feature in features:
    stat, p_value = stats.ks_2samp(first_half[feature], second_half[feature])
    status = "DRIFT DETECTED" if p_value < 0.05 else "OK"
    print(f"{feature}: p-value={round(p_value, 4)} → {status}")
```
